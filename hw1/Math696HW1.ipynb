In the cell below, take user input to get an integer and store this in a variable N (note: you'll have to google or something to find out how to get user input).

```python
N = int(input("Please enter an integer: "))
```

In the cell below, use a loop to compute $N!$.

```python
N_factorial = 1
for i in range(1, N + 1):
    N_factorial *= i
print(f"The factorial of {N} is {N_factorial}")
```

In the cell below, use a loop to compute the $N$th term in the Fibonacci sequence.

```python
N_fibonacci = [0, 1]
for i in range(2, N):
    N_fibonacci.append(N_fibonacci[i - 1] + N_fibonacci[i - 2])
print(f"The {N}th Fibonacci number is {N_fibonacci[-1]}")
```

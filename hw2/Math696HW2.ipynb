In the cell below, you are going to write a function that takes in one required parameter -- x -- and one optional parameter -- N -- with default value N = 4. It will return cos(x) computed using the first N terms of the MacLaurin series. I will put the function signature below; only modify the body.   

```python
from math import *

def my_cos(x, N = 4):
  '''
  Compute cos(x) using the first N terms of the MacLaurin series.
  params
  x:  value in radians at which we are computing cos(x)
  N:  The number of terms in the series
  returns
  cos(x)
  '''
  res = 0
  for n in range(N):
    res += ((-1)**n)*(x**(2*n))/factorial(2*n)
  return res
```

In the cell below, you will write a function that has three parameters: a, b, N and it will return a list of numbers that starts at a, ends at b (that is, b is in the list). You may not use numpy for this.

```python
def get_array(a, b, N):
  '''
  Find a list from a to b with N elements
  params
  a:  starting point
  b: ending point
  N: number of elements in the list
  returns
  a list from a to b with N elements
  '''
  ar = []
  for i in range(N):
    ar.append(a + i*(b-a)/(N-1))
  return ar
```

In the cell below you will write a function that takes in a list and returns a list that is cos(x) (using your above function) evaluated at each point of the list. So, for example, if your list is [0, pi/2, pi] then this will return [1, 0, -1]. It also takes an optional argument N that is the number of terms in the my_cos function above.  

```python
def get_cos_ar(ar, N = 5):
  '''
  Given a list ar, compute cos(ar) -- i.e. cos at each element of the list
  params
  ar: the list at which cos(x) will be computed
  N:  The number of terms in the series
  returns
  cos(ar)
  '''
  c_ar = []
  for x in ar:
    c_ar.append(my_cos(x, N))
  return c_ar
```

```python
print(f'get_cos_ar(get_array(0, pi, 7)) = {get_cos_ar(get_array(0, pi, 7))}')
```

Finally, in the cell below, you will use the above functions to compute the integral of cos on the interval [a,b]. You can decide whether to use left endpoints, right endpoints, or midpoints (or something else). The cell below is the most open ended, but when you turn it in, you should turn it in with a = 0, b = pi (so that it is easy for me to see whether your answer is "close" or not).

```python
def get_integral_cos(a, b, N = 5, M = 100):
  '''
  Compute the integral of cos(x) from a to b using the trapezoidal rule
  params
  a: starting point of the integral
  b: ending point of the integral
  N: number of terms in the series for cos(x)
  M: number of subintervals in the trapezoidal rule
  returns
  integral of cos(x) from a to b
  '''
  x_ar = get_array(a, b, M)
  y_ar = get_cos_ar(x_ar, N)
  h = (b-a)/(M-1)
  integral = 0
  for i in range(1, M-1):
    integral += y_ar[i]
  integral += (y_ar[0] + y_ar[M-1])/2
  integral *= h
  return integral
```

Running simple tests on `get_integral_cos` to make sure it works as expected

```python
print(f'get_integral_cos(0, pi) = {get_integral_cos(0, pi)}')
print(f'get_integral_cos(0, pi/2) = {get_integral_cos(0, pi/2)}')
```

````python
from math import *

def my_cos(x, N = 4):
  '''
  Compute cos(x) using the first N terms of the MacLaurin series.
  params
  x:  value in radians at which we are computing cos(x)
  N:  The number of terms in the series
  returns
  cos(x)
  '''
  res = 0
  for n in range(N):
    res += ((-1)**n)*(x**(2*n))/factorial(2*n)
  return res
````

In the cell below, you will write a function that has three parameters: a, b, h. It will return an estimate of the integral of $\cos(x)$ on the interval $[a,b]$ using left hand Riemann sums with a base of length h. The challange is that you must write this in __one line__. That is, the ```return...``` is the only line of the function (aside from the docstring above).

(Hint: if this seems daunting, then do it in several lines and then see how they can all be comined into one line).

````python
import numpy as np

def compute_integral(a, b, h):
  '''
  Compute Left Hand Riemann sum of cos(x) on interval [a,b]
  with step size h
  params
  a:  starting point
  b: ending point
  h: step size
  returns
  estimate of integral
  '''
  return np.sum(np.vectorize(my_cos)(np.arange(a, b, h))) * h
````

testing this function with various values

````python
print(f'the integral of cos from 0 to pi/2 is {compute_integral(0, pi/2, 0.01)}')
print(f'the integral of cos from pi/6 to pi/3 is {compute_integral(pi/6, pi/3, 0.01)}')
print(f'the integral of cos from 0 to pi is {compute_integral(0, pi, 0.01)}')
````

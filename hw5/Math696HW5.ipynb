In the cell below, make an expression that is a function for which there is no elementary anti--derivative (for example, $$f(x) = e^{x^2}$$ or $$f(x) = \cos(x^2)$$ are two good examples).

An example is:
```
x = sp.symbols('x')
expr = sp.exp(x)
```
EXCEPT the one above does have an elementary anti-derivative so you shouldn't do exactly that one.

````python
import sympy as sp

#Make your dictionary below.
x = sp.symbols('x')
expr = x**x
expr
````

In the cell below, use sympy to find the derivative of the expression.

````python
d_expr = sp.Derivative(expr)
display(d_expr)
d_expr.doit()
````

In the cell below, find the symbolic anti--derivative of the expression.

````python
i_expr = sp.Integral(expr)
display(i_expr)
i_expr.doit()
````

In the cell below, differentiate the anti--derivative you found above to see if it matches your original expression.

````python
d_i_expr = sp.Derivative(i_expr)
display(d_i_expr)
d_i_expr.doit()
````

In the cell below, use sympy to do a numeric integral of your expression.

````python
num_i_expr = sp.Integral(expr, (x, 0, 1))
display(num_i_expr)
num_i_expr.evalf()
````

Find the first 10 non--zero terms of the MacLaurin series of your expression and then use sympy to anti--differentiate this.  

````python
s_expr = expr.series(x, 0, 10).removeO()
display(s_expr)
num_i_s_expr = sp.Integral(s_expr, (x, 0, 1))
display(num_i_s_expr)
num_i_s_expr.evalf()
````

Finally, plot (there are several options here) the anti-derivative of the series from above.

````python
sp.plot(expr, (x, 0, 1), title="original function $x^x$")
````

````python
sp.plot(s_expr, (x, 0, 1), title="MacLaurin series of $x^x$")
````

````python
t = sp.symbols('t')
F_sym = sp.integrate(expr, (x, 0, t))
sp.plot(F_sym, (t, 0, 1), xlabel='t', ylabel='$\int_0^t x^x dx$',
        title='Cumulative integral via SymPy plot')
````

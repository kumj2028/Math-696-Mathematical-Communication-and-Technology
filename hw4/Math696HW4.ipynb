In the cell below, create a python dictionary in which the keys are names of functions and the values are the functions (for the values, use a lambda function -- example below); there should be 4 functions - you decide which ones you want to plot.

An example is:
```
fdict = {'x^2': lambda x: x**2, 'cos pi x': lambda x: np.cos(np.pi * x)}
```
but this has only 2 entries and not 4; you you'll have to add 2. Note that the
lambda stuff is new, but the line:

```
f = lambda x: x**2
```
is equivalent to:
```
def f(x):
  return x**2.
```

````python
import numpy as np
import matplotlib.pyplot as plt
x = np.linspace(-1, 1, 100)

#Make your dictionary below.
fdict = {
    'x^2': lambda x: x**2,
    'cos pi x': lambda x: np.cos(np.pi*x),
    'ln(x+2)': lambda x: np.log(x+2),
    'e^(-x^2)': lambda x: np.exp(-x**2)}
````

In the cell below, your goal is to iterate over the dictionary you made in the cell above and then plot each of those functions. For this cell, you will use plt.plot and all the graphs will be plotted on the same plot; you must label them using the keys and there must be a legend.

````python
for key in fdict:
    plt.plot(x, fdict[key](x), label=key)
    plt.legend()
    plt.show()
````

In the cell below, use the fig, axes construction to make a 2x2 plot. For this one, you will graph each function in a new plot. You must included labels and a legend as above. This one might be a little tricky, and there are several ways of doing it. Note that you basically want to iterate over two things at once: the dictionary and the axes. Look up ```zip``` online. But, like I said, there are several ways to do this and they don't all involve ```zip```.

````python
fig, axes = plt.subplots(2, 2, figsize=(10, 8))
axes = axes.flatten()

for (key, func), ax in zip(fdict.items(), axes):
    ax.plot(x, func(x), label=key)
    ax.legend()

plt.show()
````

Finally, in the cell below, use ```plt.contour``` to plot several level curves of the function $f(x,y) = x^2 - y^2$.

````python
# Create a grid of x and y values
x_vals = np.linspace(-5, 5, 100)
y_vals = np.linspace(-5, 5, 100)
X, Y = np.meshgrid(x_vals, y_vals)

# Calculate the z values (f(x,y))
Z = X**2 - Y**2

# Plot the contour lines
plt.contour(X, Y, Z, levels=20)
plt.xlabel('x')
plt.ylabel('y')
plt.title('Contour plot of $f(x,y) = x^2 - y^2$')
plt.colorbar(label='$f(x,y)$')
plt.show()
````
